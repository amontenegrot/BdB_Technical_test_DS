```python
def replace_value(df, column, old_value, new_value):
    df[column] = df[column].str.replace(old_value, new_value, regex=False)
    return df
```

# Clear data

```python
# Missing data
print(df_train.shape)
df_train.dropna(inplace=True)  # Process not necessary because there is no missing data
print(df_train.shape)
```

```python
# Irrelevant columns
cols_cat = ['Tipo_Trabajo', 'Estado_Civil', 'Educacion', 'mora',
       'Vivienda', 'Consumo', 'Contacto', 'Mes', 'Dia', 'Campana',
       'Resultado_Anterior', 'nr_employed', 'y'
       ]

print('Delete irrelevant columns:')

for col in cols_cat:  # Counting the levels in the different categorical columns
  print(f'\t column {col}: {df_train[col].nunique()} sublevels')
```

```python
# Duplicate rows
print(f'Set size before deleting repeated rows: {df_train.shape[0]}')
df_train.drop_duplicates(inplace=True)
print(f'Set size after deleting repeated rows: {df_train.shape[0]}')
```

```python
fig = go.Figure()

fig.add_trace(go.Box(x=df_train['Edad'], name='Edad',
                marker_color = 'lightseagreen'))
fig.add_trace(go.Box(x=df_train['Dias_Ultima_Camp'], name='Dias_Ultima_Camp',
                marker_color = 'lightseagreen'))
fig.add_trace(go.Box(x=df_train['No_Contactos'], name='No_Contactos',
                marker_color = 'lightseagreen'))
fig.add_trace(go.Box(x=df_train['emp_var_rate'], name='emp_var_rate',
                marker_color = 'indianred'))
fig.add_trace(go.Box(x=df_train['cons_price_idx'], name='cons_price_idx',
                marker_color = 'indianred'))
fig.add_trace(go.Box(x=df_train['cons_conf_idx'], name='cons_conf_idx',
                marker_color = 'indianred'))
fig.add_trace(go.Box(x=df_train['euribor3m'], name='euribor3m',
                marker_color = 'indianred'))

fig.show()
```

```python
# Represent in lowercase only if column is categorical
cols_cat = ['Tipo_Trabajo', 'Estado_Civil', 'Educacion', 'mora',
            'Vivienda', 'Consumo', 'Contacto', 'Mes', 'Dia', 'Campana',
            'Resultado_Anterior', 'nr_employed'
]

for column in df_train.columns:
    if column in cols_cat:
        df_train[column] = df_train[column].str.lower()
```

```python
for column in df_train.columns:
    print(f"Repeated values in <{column}>:", df_train[column].isnull().sum())
```

```python
# Manual label encoding
encode = {'si': 1, 'no': 0}
df_train = df_train.replace(encode).astype(int)
df_test = df_test.replace(encode).astype(int)
```

```python
# Define the values to be replaced in a dictionary
replace_dict = {
    'Estado_Civil': [('single', 'soltero'), ('divorced', 'divorciado')]
}

# Iterate over the columns and apply the function
for column, replacements in replace_dict.items():
    for old_value, new_value in replacements:
        df_train = replace_value(df_train, column, old_value, new_value)
```
